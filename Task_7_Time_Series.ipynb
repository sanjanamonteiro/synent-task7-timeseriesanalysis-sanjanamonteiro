# Time Series Analysis




```python
# Import libraries for time series analysis
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load stock dataset
df = pd.read_csv("/content/AAPL.csv")
```

```python
# Display first 5 rows
df.head()
```

```python
# Display dataset information
df.info()
```

```python
# Convert Date column into datetime format
df['Date'] = pd.to_datetime(df['Date'])
```

```python
# Sort records by date
df = df.sort_values('Date')
```

```python
# Create stock closing price trend graph
plt.figure(figsize=(12,5))

plt.plot(df['Date'], df['Close'])

plt.title("Apple Stock Closing Price Trend")
plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.show()
```

```python
# Calculate 30-day moving average
df['MA30'] = df['Close'].rolling(window=30).mean()
```

```python
# Compare actual prices with moving average
plt.figure(figsize=(12,5))
plt.plot(df['Date'], df['Close'], label='Close Price')
plt.plot(df['Date'], df['MA30'], label='30-Day Moving Average')
plt.title("Stock Price vs Moving Average")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.show()
```

1. Stock prices show long-term growth trends.
2. Moving average helps identify overall market direction.
3. Short-term fluctuations are reduced using moving averages.
4. Time series analysis helps understand future market behavior.
5. Historical trends can support forecasting decisions.
